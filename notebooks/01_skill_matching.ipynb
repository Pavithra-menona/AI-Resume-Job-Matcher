```python
resume_skills = [
    "Python",
    "HTML",
    "CSS",
    "JavaScript",
    "Pandas",
    "Git"
]

job_skills = [
    "Python",
    "Pandas",
    "SQL",
    "Machine Learning",
    "Git",
    "Docker"
]

matched_skills = set(resume_skills) & set(job_skills)
missing_skills = set(job_skills) - set(resume_skills)

print("Matched Skills:")
print(matched_skills)

print("\nMissing Skills:")
print(missing_skills)
```

```python
resume_text = """
I am a Computer Science Engineering student with experience in
Python, HTML, CSS, JavaScript, Pandas and Git.
I have also worked on web development projects.
"""

job_description = """
We are looking for a Software Developer Intern.
Required skills include Python, SQL, Pandas, Git,
Machine Learning and Docker.
Experience with web development is preferred.
"""

print("Resume:")
print(resume_text)

print("\nJob Description:")
print(job_description)
```

```python
skills_database = [
    "Python",
    "Java",
    "C",
    "C++",
    "HTML",
    "CSS",
    "JavaScript",
    "React",
    "Django",
    "Flask",
    "SQL",
    "Pandas",
    "NumPy",
    "Machine Learning",
    "Deep Learning",
    "Git",
    "GitHub",
    "Docker",
    "AWS",
    "Figma"
]

print("Number of skills in database:", len(skills_database))
```

```python
resume_lower = resume_text.lower()

resume_found_skills = []

for skill in skills_database:
    if skill.lower() in resume_lower:
        resume_found_skills.append(skill)

print("Skills found in resume:")
print(resume_found_skills)
```

```python
job_lower = job_description.lower()

job_found_skills = []

for skill in skills_database:
    if skill.lower() in job_lower:
        job_found_skills.append(skill)

print("Skills required by the job:")
print(job_found_skills)
```

```python
matched_skills = set(resume_found_skills) & set(job_found_skills)
missing_skills = set(job_found_skills) - set(resume_found_skills)

match_percentage = (
    len(matched_skills) / len(job_found_skills)
) * 100

print("========== JOB MATCH REPORT ==========")

print("\nMatched Skills:")
for skill in matched_skills:
    print("✓", skill)

print("\nMissing Skills:")
for skill in missing_skills:
    print("✗", skill)

print(f"\nMatch Percentage: {match_percentage:.1f}%")
```

```python
skill_weights = {
    "Python": 3,
    "SQL": 3,
    "Machine Learning": 3,
    "Pandas": 2,
    "Git": 2,
    "Docker": 2,
    "HTML": 1,
    "CSS": 1,
    "JavaScript": 1,
    "Figma": 1
}

print(skill_weights)
```

```python
total_weight = 0
matched_weight = 0

for skill in job_found_skills:
    weight = skill_weights.get(skill, 1)
    total_weight += weight

    if skill in resume_found_skills:
        matched_weight += weight

weighted_match = (matched_weight / total_weight) * 100

print(f"Weighted Job Match: {weighted_match:.1f}%")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
resume_sentence = """
I have experience in data analysis using Python and Pandas.
"""

job_sentence = """
The candidate should have knowledge of Python-based data analytics
and experience working with Pandas.
"""

vectorizer = TfidfVectorizer()

vectors = vectorizer.fit_transform([
    resume_sentence,
    job_sentence
])

similarity = cosine_similarity(vectors[0], vectors[1])[0][0]

print(f"Similarity Score: {similarity:.2f}")
```

```python
documents = [
    resume_text,
    job_description
]

vectorizer = TfidfVectorizer(
    stop_words="english"
)

vectors = vectorizer.fit_transform(documents)

similarity = cosine_similarity(
    vectors[0],
    vectors[1]
)[0][0]

print(f"Resume-Job Text Similarity: {similarity:.2f}")
```

```python
skill_score = weighted_match
nlp_score = similarity * 100

final_score = (
    0.70 * skill_score +
    0.30 * nlp_score
)

print(f"Skill Score: {skill_score:.1f}%")
print(f"NLP Similarity: {nlp_score:.1f}%")
print(f"Final Match Score: {final_score:.1f}%")
```

```python
def extract_skills(text, skills_data):
    text = text.lower()
    found_skills = []

    for skill, details in skills_data.items():
        for alias in details["aliases"]:
            if alias.lower() in text:
                found_skills.append(skill)
                break

    return found_skills
```

```python
detected_resume_skills = extract_skills(
    real_resume_text,
    skills_data
)

print("Skills detected in resume:")

for skill in detected_resume_skills:
    print("✓", skill)
```

```python
detected_job_skills = extract_skills(
    real_job_text,
    skills_data
)

print("Skills required by job:")

for skill in detected_job_skills:
    print("•", skill)
```
